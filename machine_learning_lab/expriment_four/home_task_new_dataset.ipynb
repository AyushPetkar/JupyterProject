# Heart Disease Classification – Home Task (New Dataset)

## About Dataset – Heart Disease UCI Dataset

**1. Dataset Name/Title:** Heart Disease UCI Dataset

**2. Source:** UCI Machine Learning Repository. Available on Kaggle via: https://www.kaggle.com/datasets/johnsmith88/heart-disease-dataset

**3. Purpose:** To predict whether a patient has heart disease (1) or not (0) based on clinical measurements. This binary classification problem helps in early detection and medical decision-making.

**4. Size/Dimensions:**
- Total Records: 1,025 samples
- Features: 13 clinical attributes (all numeric) + 1 target variable
- Structure: 1,025 rows × 14 columns
- Class Distribution: ~526 positive (heart disease), ~499 negative – roughly balanced

**5. Description of Features:**

| Feature | Description | Data Type |
|---------|-------------|-----------|
| age | Age of the patient | Numeric |
| sex | Sex (1 = male, 0 = female) | Binary |
| cp | Chest pain type (0–3) | Numeric |
| trestbps | Resting blood pressure (mm Hg) | Numeric |
| chol | Serum cholesterol (mg/dl) | Numeric |
| fbs | Fasting blood sugar > 120 mg/dl | Binary |
| restecg | Resting electrocardiographic results | Numeric |
| thalach | Maximum heart rate achieved | Numeric |
| exang | Exercise induced angina (1 = yes) | Binary |
| oldpeak | ST depression induced by exercise | Numeric |
| slope | Slope of peak exercise ST segment | Numeric |
| ca | Number of major vessels colored by fluoroscopy | Numeric |
| thal | Thalassemia type (0–3) | Numeric |

**6. Target Variable:**
| Class Label | Description |
|-------------|-------------|
| 0 | No heart disease |
| 1 | Heart disease present |

**Additional Notes:**
- All features are numeric — no label encoding required (simpler than Bank Marketing)
- Dataset is roughly balanced, similar to Iris
- Well-suited for Decision Tree classification

```python
import os

import kagglehub
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
```

## Step 1: Download Dataset from Kaggle

```python
# Download latest version from Kaggle
path = kagglehub.dataset_download("johnsmith88/heart-disease-dataset")
print("Path to dataset files:", path)
```

```python
# List available files in the downloaded dataset
print(os.listdir(path))
```

```python
# Load the dataset
df = pd.read_csv(os.path.join(path, "heart.csv"))
print("Dataset loaded successfully!")
```

## Step 2: Exploratory Data Analysis (EDA)

```python
# Check dataset shape
df.shape
```

```python
# View first 5 rows
df.head()
```

```python
# View last 5 rows
df.tail()
```

```python
# Statistical summary
df.describe()
```

```python
# Check for missing/null values
df.isnull().sum()
```

```python
# Check target variable distribution
print("Target Variable Distribution:")
print(df['target'].value_counts())
print("\nPercentage distribution:")
print(df['target'].value_counts(normalize=True) * 100)
```

## Step 3: Data Visualization

```python
# Plot age distribution by target (mirrors sepal length plot in Experiment 4)
plt.figure(figsize=(10, 6))
for label in df['target'].unique():
    subset = df[df['target'] == label]
    plt.hist(subset['age'], alpha=0.5, label=f'target={label}', bins=20)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Age Distribution by Heart Disease Status')
plt.legend()
plt.show()
```

```python
# Correlation heatmap to explore feature relationships
import seaborn as sns

plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Heatmap')
plt.show()
```

## Step 4: Create Dependent and Independent Variables

```python
# Split features and target
X = df.drop('target', axis=1)
y = df['target']
```

```python
# Check shape of X and y
X.shape
```

```python
y.shape
```

## Step 5: Train-Test Split

```python
# Split data into training and testing sets (same parameters as Experiment 4)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=4)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
y_test.shape
```

## Step 6: Model Training – Decision Tree Classifier

```python
# Train the Decision Tree Classifier (same as Experiment 4)
model = DecisionTreeClassifier(criterion='entropy')
model.fit(X_train, y_train)
```

```python
# Predict on test set
y_pred = model.predict(X_test)
```

## Step 7: Model Evaluation

```python
# Accuracy score
accuracy_score(y_test, y_pred)
```

```python
# Precision score
precision_score(y_test, y_pred, average='weighted')
```

```python
# Confusion matrix
confusion_matrix(y_test, y_pred)
```

```python
# F1 score
f1_score(y_test, y_pred, average='weighted')
```

```python
# Model score on test set
model.score(X_test, y_test)
```

```python
# Feature importances
model.feature_importances_
```

```python
# Display feature importances as a labelled series for readability
feature_importance_series = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
print(feature_importance_series)
```

## Step 8: Predict on a Single Test Sample

```python
first_test_input = X_test.iloc[7]
actual_output = y_test.iloc[7]
predicted_output = model.predict(first_test_input.values.reshape(1, -1))

print(f"Predicted Output: {predicted_output}")
print(f"Actual Output: {actual_output}")
assert predicted_output == actual_output
```

## Step 9: Plot the Decision Tree

```python
plt.figure(figsize=(20, 10))
plot_tree(
    model,
    filled=True,
    feature_names=X.columns,
    class_names=[str(c) for c in model.classes_],
    max_depth=3  # limit depth for readability
)
plt.title('Decision Tree – Heart Disease Classification')
plt.show()
```

## Step 10: Cross-Validation

```python
cv_model = DecisionTreeClassifier(random_state=0)
cv_scores = cross_val_score(cv_model, X, y, cv=5)
print("Cross-validation scores:", cv_scores)
print("Average cross-validation score:", cv_scores.mean())
print("Standard deviation:", cv_scores.std())
```
